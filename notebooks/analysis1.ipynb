```python
gender_counts = df['Gender'].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(
    gender_counts, 
    labels=gender_counts.index, 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=sns.color_palette('pastel')
)
plt.title('Gender Distribution (Pie Chart)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('../charts/gender_pie_chart.png', dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

# Fixed line: assign x variable to hue and disable redundant legend
sns.countplot(
    data=df, 
    x="Investment_Interest", 
    hue="Investment_Interest", 
    palette="viridis", 
    legend=False
)

plt.title("Investment Interest Among Survey Participants")
plt.xlabel("Investment Category")
plt.ylabel("Number of Participants")
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig("../charts/investment_interest.png", dpi=300)
plt.show()
```

```python
# Identify experience column (index 7 based on your CSV layout)
col_exp = df.columns[7]
exp_counts = df[col_exp].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(
    exp_counts,
    labels=exp_counts.index,
    autopct='%1.1f%%',
    startangle=140,
    pctdistance=0.8,
)


# Central circle cutout for the donut effect

plt.title('Stock Market Experience Distribution', fontsize=14)
plt.tight_layout()
plt.savefig('../charts/experience_donut_chart.png', dpi=300)
plt.show()
```

```python
col_dash = 'Dashboard_Useful' if 'Dashboard_Useful' in df.columns else 'Dashboard Useful'
dash_counts = df[col_dash].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(
    dash_counts, 
    labels=dash_counts.index, 
    autopct='%1.1f%%', 
    startangle=90, 
    pctdistance=0.8,
    colors=sns.color_palette('Set2')
)

# Draw a circle in the middle to create the donut effect


plt.title('Perception of Dashboard Usefulness (Donut Chart)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('../charts/dashboard_donut_chart.png', dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(
    data=df,
    x="Occupation",
    hue="Occupation",
    palette="mako",
    legend=False
)
plt.title("Participant Distribution by Occupation")
plt.xlabel("Occupation")
plt.ylabel("Count")
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig("../charts/occupation_distribution.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['Age'], bins=8, kde=True, color='#2b5c8f', edgecolor='black')
plt.title('Age Distribution of Survey Participants', fontsize=14, fontweight='bold')
plt.xlabel('Age', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.tight_layout()
plt.savefig('../charts/age_distribution.png', dpi=300)
plt.show()
```

```python
col_info = 'Most_Useful_Information' if 'Most_Useful_Information' in df.columns else 'Most Useful Information'

plt.figure(figsize=(9, 5))
order_info = df[col_info].value_counts().index
sns.countplot(data=df, y=col_info, hue=col_info, order=order_info, palette='magma', legend=False)
plt.title('Most Useful Information Source', fontsize=14, fontweight='bold')
plt.xlabel('Count', fontsize=12)
plt.ylabel('Information Category', fontsize=12)
plt.tight_layout()
plt.savefig('../charts/most_useful_information.png', dpi=300)
plt.show()
```

```python
col_invest = 'Investment_Interest' if 'Investment_Interest' in df.columns else 'Investment Interest'

# Create a cross-tabulation table
cross_tab = pd.crosstab(df['Occupation'], df[col_invest])

plt.figure(figsize=(8, 5))
sns.heatmap(cross_tab, annot=True, fmt='d', cmap='YlGnBu')
plt.title('Occupation vs Preferred Investment (Heatmap)', fontsize=14, fontweight='bold')
plt.xlabel('Investment Preference', fontsize=12)
plt.ylabel('Occupation', fontsize=12)
plt.tight_layout()
plt.savefig('../charts/occupation_vs_investment_heatmap.png', dpi=300)
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.boxplot(data=df, x='Occupation', y='Age', hue='Occupation', palette='Set3', legend=False)
plt.title('Age Distribution Across Occupations', fontsize=14, fontweight='bold')
plt.xlabel('Occupation', fontsize=12)
plt.ylabel('Age', fontsize=12)
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig('../charts/age_occupation_boxplot.png', dpi=300)
plt.show()
```
